```python
from langchain_google_genai import ChatGoogleGenerativeAI
from typing import TypedDict
from langchain_core.messages import SystemMessage, HumanMessage
import os
```

```python

#state of the nodes 
class NodeState(TypedDict):
    title:str
    message:str
```

```python
from langgraph.graph import StateGraph, START, END
```

```python
llm=ChatGoogleGenerativeAI(model="gemini-2.5-flash",  # Option A: Stable updated 1.5 engine
    # model="gemini-2.5-flash",     # Option B: Recommended faster/newer generation
    temperature=0.7)
def generate_title_blog(state:NodeState)->dict:
    input_title=state['title'].strip()
    if not input_title:
        print("Write Belog ON MY name")
    prompt=[ 
        SystemMessage(content=("You are an expert blog writer. Write a comprehensive, "
    "engaging, and well-structured blog post based on the given title. Include an introduction, subheadings, and a conclusion.")),
    HumanMessage(content= f"Write a complete blog post for this title: {input_title}")
    ]
    response=llm.invoke(prompt)
    return {"title":input_title, "message":response.content}
```

```python
app=StateGraph(NodeState)
app.add_node("Blog_Generation", generate_title_blog)
app.add_edge(START, "Blog_Generation")
app.add_edge("Blog_Generation",END)
app=app.compile()
app
```

```python
result=app.invoke({"title":"The Future of Artificial Intelligence: Trends and Predictions for 2024 and Beyond"})
print(result["message"])
```
