```python
from pydantic import BaseModel
from typing import TypedDict,Literal
from langgraph.graph import StateGraph, START,END
```

```python
class NodeState(BaseModel):
    num:int
```

```python
def input(state:NodeState)->dict:
    return {"num":state.num}
def router(state:NodeState)->Literal["positive","negative"]:
    if state.num >= 0:
        print("positive number")
        return "positive"
    else:
        print("negative number")
        return "negative"

def positive(state:NodeState)->dict:
    return {"num":state.num}
def negative(state:NodeState)->dict:
    return {"num":state.num}

```

```python
graph=StateGraph(NodeState)
graph.add_node("Input",input)
graph.add_node("pos",positive)
graph.add_node("neg",negative)
graph.add_edge(START,"Input")
graph.add_conditional_edges("Input",router,{
"positive":"pos",
"negative":"neg"
})
graph.add_edge("pos",END)
graph.add_edge("neg",END)
app=graph.compile()
result=app.invoke({
    "num":5
})
print(result)
```
