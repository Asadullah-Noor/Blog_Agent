```python
# ==============================
# Agent ID vs Thread ID — TEST
# ==============================

# Application-level Agent IDs
agents = {
    "agent-001": "Research Agent",
    "agent-002": "Writer Agent"
}

# Simulated checkpoint storage
checkpoints = {}


def run_agent(agent_id, thread_id, message):

    # Check whether this thread already has state
    previous_state = checkpoints.get(thread_id, [])

    # Add current message
    previous_state.append(message)

    # Save state against THIS thread
    checkpoints[thread_id] = previous_state

    print("\n-----------------------------")
    print("Agent ID  :", agent_id)
    print("Thread ID :", thread_id)
    print("State     :", checkpoints[thread_id])
    print("-----------------------------")


# ==================================
# 1. Agent-001, Thread-001
# ==================================

run_agent(
    "agent-001",
    "thread-001",
    "Research Lenovo"
)


# ==================================
# 2. SAME Agent, SAME Thread
# ==================================

run_agent(
    "agent-001",
    "thread-001",
    "Now research its revenue"
)


# ==================================
# 3. SAME Agent, NEW Thread
# ==================================

run_agent(
    "agent-001",
    "thread-002",
    "Research Apple"
)


# ==================================
# 4. DIFFERENT Agent, NEW Thread
# ==================================

run_agent(
    "agent-002",
    "thread-003",
    "Write an article about AI"
)
```
