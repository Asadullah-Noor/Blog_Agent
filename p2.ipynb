```python
from pydantic import BaseModel
from typing import TypedDict,Literal
from langgraph.graph import StateGraph,START,END
```

```python
def Node(TypedDict):
    a:int
    b:int
def Nodestate(BaseModel):
    a:int
    b:int
    operation:list[str]
```

```python
def input(state:Nodestate)->dict:
    return {"a":state.a,"b":state.b}
def addition(state:Nodestate)->dict:
    return {state.a+state.b}
def subtraction(state:Nodestate)->dict:
    return {state.a-state.b}
def multiply(state:Nodestate)->dict:
    return {state.a*state.b}
def division(state:Nodestate)->dict:
    return {state.a/state.b}
```

```python
def checknode(state:Nodestate)->Literal['addition','subtraction','multiply','division']:
    if state['operation']=='addition':
        return {"operation":addition}
    elif state['operation']=='subtraction':
        return {"operation":subtraction}
    elif state['operation']=='multiply':
        return {"operation":multiply}
    else :
        return {'operation':division}

```

```python
graph=StateGraph(Nodestate)
graph.add_node('Input',input)
graph.add_node('Addition',addition)
graph.add_node('Subtraction',subtraction)
graph.add_node('Multiply',multiply)
graph.add_node('Division',division)
graph.add_edge(START,'Input')
graph.add_conditional_edges('Input',
               checknode,{
                "addition":'Addition',
                'subtraction':'Subtraction',
                'multiply':'Multiply',
                'division':'Division'
               })
graph.add_edge('Addition',END)
graph.add_edge('Subtraction',END)
graph.add_edge('Multiply',END)
graph.add_edge('Division',END)
graph.compile()
result=graph.invoke({
    'a':4,
    'b':5,
    'operation':'multiply'
})
print(result)
```
